### Inheritance

```python
# One parent one child

class Animal:
    def roar(self):
        print("animal roars")
    
class baby(Animal):
    def cry(self):
        print("baby cries")

a = baby()
a.cry()
a.roar()
```

### We are doing parent child class

```python
class parent:
    def show(self):
        print("I am parent class!")

class child(parent):
    def display(self):
        print("I am child!")

a=child()
a.show()
a.display()
```

### Multilevel Inheritance

```python
class grandfather:
    def show(self):
        print("I am grandfather")

class father(grandfather):
    def display(self):
        print("I am father")

class child(father):
    def generate(self):
        print("I am child")

x = child()
x.generate()
x.display()
x.show()
```

### Multiple Inheritance

```python
class father:
    def show(self):
        print("I am father")

class mother:
    def display(self):
        print("I am mother")

class child(father, mother):
    def generate(self):
        print("I am child")

x = child()
x.generate()
x.display()
x.show()
```
