### Inheritance

Inheritance is a concept where a child class acquires the properties and methods of a parent class.

It helps in code reusability and better organization.

```python
# One parent one child

class Animal:
    def roar(self):
        print("animal roars")
    
class baby(Animal):
    def cry(self):
        print("baby cries")

a = baby()
a.cry()
a.roar()
```

### We are doing parent child class

```python
class parent:
    def show(self):
        print("I am parent class!")

class child(parent):
    def display(self):
        print("I am child!")

a=child()
a.show()
a.display()
```

### Multilevel Inheritance

```python
class grandfather:
    def show(self):
        print("I am grandfather")

class father(grandfather):
    def display(self):
        print("I am father")

class child(father):
    def generate(self):
        print("I am child")

x = child()
x.generate()
x.display()
x.show()
```

### Multiple Inheritance

```python
class father:
    def show(self):
        print("I am father")

class mother:
    def display(self):
        print("I am mother")

class child(father, mother):
    def generate(self):
        print("I am child")

x = child()
x.generate()
x.display()
x.show()
```

### Hierarchial Inheritance

When one parent class is inherited by multiple child classes, it is called hierarchical inheritance.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

class Cat(Animal):
    def meow(self):
        print("Cat meows")

# Creating objects
d = Dog()
c = Cat()

d = Dog()
c = Cat()

d.sound()
d.bark()

c.sound()
c.meow()



```

#### Polymorphism

Polymorphism means “many forms.”
In Python, it allows the same function or method name to behave differently for different objects


```python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("Cat meows")

d = Dog()
c = Cat()

d.sound()   # sound method in Dog class overrides sound method in Animal class
c.sound()
```

### Method Overriding

```python
class Animal:
    def sound(self):
        print("animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("dog barks")

a=Dog()
b=Animal()

a.sound()
b.sound()
```
