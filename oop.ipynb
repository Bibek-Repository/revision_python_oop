### We are learning Constructor in Python

```python
class Person:
    def __init__(self,name, age):
        self.name = name
        self.age = age
    
p=Person("Ananda", 21)
```

```python
class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.sound = sound
```

### Super Keyword in python

The super() keyword is used to call methods or constructors of the parent class from a child class.

```python
class employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print(f"name of the employee is {self.name} and salary is {self.salary}")

class developer(employee):
    def __init__(self,name, salary, bonus):
        super().__init__(name, salary)   # this function will automatically pass arguments to the parent class
        super().display()
        self.name = name
        self.salary = salary
        self.bonus = bonus
    
    def display(self):
        print(f"developer name: {self.name}, salary of the developer is {self.salary} and bonus of the developer is {self. bonus} ")

class manager(employee):
    def __init__(self, name, salary, age):
        super().__init__(name, salary)
        super().display()
        self.name=name
        self.salary=salary
        self.age = age

    def display(self):
        print(f" name of the manager is {self.name}, salary of the manager is {self.salary}, age of the manager is {self.age}")

p1=developer("Bibek", 100000, 20000)
p1.display()
p2 = manager("Jack", 500000, 45)
p2.display()
```
