```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def display(self):
        print(f"Name of the person is: {self.name} and age is: {self.age}")
p1 = Person("John", 36)
print(p1.name)
print(p1.age)
p1.display()
p2 = Person("Bibek", 25)
p2.name
p2.display()
```

### Default Values in __init__()

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def display(self):
        print(f"Name of the person is: {self.name} and age is: {self.age}")
p1 = Person("Ananda", 25)
print(p1.name)
print(p1.age)
p1.display()
p2 = Person("Ram", 30)
print(p2.name)
print(p2.age)
p2.display()
```
