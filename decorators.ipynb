### Decorators

```python
def simple_decorator(func):
    def wrapper(*args):     # *args is used for any number of argument receive
        print("good morning")
        result=func(*args)
        print("Thanks for using the function")
    return wrapper
@simple_decorator
def display(name):
    print(f"name is  {name}")

display("Bibek")

@simple_decorator
def show(name):
    print(f"hello i am {name}")

@simple_decorator
def add(a,b):
    print(a+b)

add(2,3)

show("ronaldo")
```

### Practise

```python
def simple_decorate(func):
    def wrapper(*args):
        print("this is mathematical operation")
        result = func(*args)
        print("Mathematical operation completed")
    return wrapper

@simple_decorate
def divide(a,b):
    print(f"division: {a+b}")

@simple_decorate
def diff(a,b):
    print(f"difference: {a-b}")

@simple_decorate
def multiply(a,b,c):
    print(f"multiplication: {a*b*c}")

divide(4,2)
diff(5,3)
multiply(1,2,3)







```
