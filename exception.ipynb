### Exception Handling in Python

Exception handling is used to handle runtime errors so that the program does not crash and runs smoothly.
Exception handling is a mechanism to handle runtime errors using try, except, else, and finally.

1. Why Exception Handling is Important? 

    Prevents program crash

    Handles errors gracefully

    Improves user experience

    Helps debugging

2. Common Built-in Exceptions 
    Exception	                    Reason
    ZeroDivisionError	            Divide by zero
    ValueError	                    Invalid value
    TypeError	                    Wrong data type
    IndexError	                    List index out of range
    KeyError	                    Dictionary key missing
    FileNotFoundError	            File not found

Syntax:
try:
    # code that may cause error
except:
    # code to handle error

```python
# Example 1: Handling Division by Zero

try:
    a = 10
    b = 0
    print(a / b)
except:
    print("Cannot divide by zero")
```

```python
# Example 2: Handling Specific Exception

try:
    x = int("abc")
except ValueError:
    print("Invalid number")
```

```python
# Example 3: Multiple Exceptions

try:
    a = int(input("Enter number: "))
    b = int(input("Enter number: "))
    print(a / b)
except ZeroDivisionError:
    print("Division by zero is not allowed")
except ValueError:
    print("Please enter valid numbers")

```

```python
# else Block

# Executed only if no exception occurs.

try:
    print("No error here")
except:
    print("Error occurred")
else:
    print("Program ran successfully")
```

```python
# finally Block
# Executed always, whether an exception occurs or not.

try:
    print(10 / 2)
except:
    print("Error")
finally:
    print("This always executes")
```

### Custom error handling

A custom (user-defined) exception is created by inheriting the built-in Exception class.
Exception is the parent class of most Python exceptions.

Basic Syntax
class CustomError(Exception):
    pass


```python
# Example 1: Simple Custom Exception

class MyError(Exception):
    pass

try:
    raise MyError("This is a custom exception")
except MyError as e:
    print(e)
```

```python
# Example 2: Custom Exception with Condition

class AgeError(Exception):
    pass

age = -3

try:
    if age < 0:
        raise AgeError("Age cannot be negative")
    print("Valid age")
except AgeError as e:
    print("Error:", e)

```
